# ****CATATAN PEMAHAMAN DAN IDEASI****

---

Terkait ideasi, lebih baik difokuskan ke Enterprise Data Management.

Karena, tata kelola data sangat diutamakan:

1. kualitas data
2. dokumentasi
3. validasi
4. reproducibility

Performa model bukan segalanya. Yang dinilai juga kemampuan menjelaskan metodologi, analisis, dan dampak. Utamakan:

1. Masalah tajam
2. Data kredibel
3. Evaluasi yang jujur



****STRATEGI UNTUK JUARA****

1. Pilih masalah yang spesifik, lokal, dan datanya kuat. Hindari topik yang pasti dipakai banyak tim (seperti, prediksi diabetes, harga rumah, churn, sentimen generik). Yang dibutuhkan adalah masalah Indonesia yang konkret dengan pengguna keputusan yang jelas, misalnya PEMDA, BPBD, atau Dinas Kesehatan.

2. Bangun nilai pada datanya. Dataset yang menggabungkan beberapa sumber resmi (seperti, BMKG, BPS, NASA FIRMS, Satu Data Indonesia, Kemenkes, World Bank, dan Citra Satelit) lalu diolah dengan feature engineering yang bermakna akan terlihat jauh lebih kuat daripada satu CSV siap pakai. Data spasial-temporal biasanya paling kaya untuk itu.

Rigor ML yang membedakan:

1. Buat baseline sederhana dulu, lalu model yang lebih kompleks, supaya terbukti kompleksitasnya layak.
2. Validasi yang sesuai karakter data, misalnya split temporal atau spasial, bukan random split asal-asalan.
3. Tuning sistematis (mis. Optuna) dan ablation study untuk menunjukkan fitur atau komponen mana yang benar-benar berkontribusi
4. Metrik yang selaras dengan keputusan nyata, termasuk analisis error per segmen (daerah, kelompok) dan kalibrasi.
5. Explainability (SHAP atau sejenisnya) yang diterjemahkan ke insight kebijakan, bukan sekadar plot.
6. Bagian keterbatasan dan etika yang jujur. Juri sering terkesan pada tim yang tahu kelemahannya.

>Ubah model menjadi solusi. Untuk skor inovasi dan dampak, model sebaiknya dibungkus sebagai alat bantu keputusan: prototipe dashboard atau peta risiko, skenario penggunaan, dan estimasi dampak yang terukur (misalnya berapa wilayah yang bisa diprioritaskan dengan sumber daya terbatas).

**Kandidat arah studi kasus**

Ini baru daftar awal untuk disaring, bukan keputusan:

1. Prediksi risiko kebakaran hutan dan lahan (titik panas, cuaca, tipe lahan gambut) --> SDG 13 dan 15 --> Spasial-temporal, data terbuka kaya, dampak nyata

2. Pemetaan kemiskinan atau kerentanan wilayah dengan citra satelit (cahaya malam, tutupan lahan) + data BPS --> SDG 1 dan 10 --> Pendekatan inovatif, relevan untuk intervensi

3. Risiko stunting atau gizi tingkat kabupaten/kota (SSGI, Susenas, sanitasi, akses air) --> SDG 2, 3, dan 6 --> Prioritas nasional, tapi data individual terbatas

4. Peramalan harga pangan dan kerentanan pasokan (PIHPS, cuaca, panen) --> SDG 2 --> Time series, dampak ekonomi jelas

5. Kualitas udara dan kesehatan perkotaan --> SDG 3 dan 11 --> Data sensor tersedia, mudah didemokan

****RENCANA KERJA****

1. Riset awal: eksplorasi data (Cek kelengkapan, lisensi, ukuran, dan data leakage)
2. Pemilihan topik: keputusan lewat matriks penilaian, bukan selera.
3. Pembangunan data: pengumpulan, pembersihan, data dictionary, dan feature engineering terdokumentasi.
4. Modeling dan evaluasi: baseline, model lanjutan, tuning, ablation, error analysis, explainability.
5. Solusi dan dampak: prototipe, estimasi dampak, dan narasi SDG.
6. Pengemasan: repo dan notebook yang bersih serta reproducible.

### ****IDEASI****

REKOMENDASI: Karhutla dengan framing yang berbeda.

Risikonya terus terang adalah topik ini ramai dipakai peserta lain. Jadi kita tidak boleh sekadar “memprediksi titik panas”. Ini bentuk yang menurutku bisa membedakan dari yang lain.

1. Masalah keputusan, bukan sekadar prediksi. Tim patroli dan pemadam terbatas. Pertanyaannya: dari sekian ribu sel wilayah, mana yang harus diprioritaskan dalam 1-2 minggu ke depan? Metrik utamanya adalah berapa persen kejadian yang tertangkap di 10% wilayah teratas, bukan akurasi.

2. Validasi prospektif yang jujur. Karena musim kebakaran 2026 baru saja lewat, kita bisa melatih model dengan data sebelum 2026 lalu mengujinya pada 2026 sebagai uji ke depan yang benar-benar asli. Pergeseran pola ke wilayah baru juga menjadi ujian distribution shift. Sedikit tim yang akan punya uji seperti ini.

3. Ground truth yang dibahas kritis. Kita bandingkan beberapa definisi label (titik panas dengan confidence tinggi, area terbakar, dan sebagainya) dan tunjukkan bagaimana pilihan label memengaruhi hasil. Ini sekaligus menjawab bobot Pengolahan Data dan kesan EDM Lab soal kualitas data.

4. Sudut kebakaran berulang pada lahan yang pernah terbakar (gambut). Ini lebih tajam dan lebih bermakna kebijakan.

5. Rigor model: baseline klimatologi dan logistic regression, lalu gradient boosting, kemudian pembanding spatio-temporal (misalnya ConvLSTM) untuk membuktikan kompleksitas tambahan layak atau tidak. Ditambah spatial block CV, kalibrasi probabilitas, SHAP per wilayah (gambut Sumatra/Kalimantan berbeda dari savana NTT dan Papua), dan ablation fitur
